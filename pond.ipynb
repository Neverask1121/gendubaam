```python
## Splitting array and other stuff

import numpy as np
arr = np.array([[1,2,3,], [2,43,56], [4,6,8], [45,54,5], [453,453,5]])
newarr = np.array_split(arr, 3, axis =1)
print(arr)
print(newarr)
```

```python
newarr = np.hsplit(arr, 3)
print(newarr)
```
